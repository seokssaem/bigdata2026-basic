## 불(bool) 인덱싱

```python
w = df['temp'] >= 30
df[w] # 참(True)인것만 행나온다
```

```python
w
```

```python
df[df['temp'] >= 30]
```

### 조건에 맞는 데이터 추출

```python
# 최고로 더웠던 날의 모든 정보(행)
w = df['temp'] == df['temp'].max()
df[w]
```

```python
df[df['temp'] == df['temp'].max()]
```

## 조건이 2개 이상 --> 그리고 &, 또는 |

```python
# 기온이 30도 이상이고, 최대풍속이 9이상인 데이터

w = (df['temp'] >= 30) & (df['max_wind'] >= 9)
df[w]
```

```python
# 기온이 30도 이상이거나, 최대풍속이 9이상인 데이터

w = (df['temp'] >= 30) | (df['max_wind'] >= 9)
df[w]
```

# 판다스 결측데이터 처리

## isna()  --> 값이 nan(결측치)이라면 True

```python
# 최대풍속이 측정되지 않은 모든 데이터 출력

df['max_wind'].isna()
```

```python
df[df['max_wind'].isna()]
```

## 데이터프레임명.isnull().sum()  --> 컬럼별 결측데이터 개수 

- 0 False 거짓
- 1 True 참

```python
df.isnull()
```

```python
df.isnull().sum()  # 컬럼별
```

```python
df.isnull().sum().sum()  # 전체 데이터프레임
```

## 결측데이터 삭제하기 --> dropna()

```python
df.dropna()  # 결측치를 제거한 상태를 한번 보는 용도
```

```python
df # 원본 데이터프레임의 결측치가 제거되지 않았다
```

```python
df2 = df.dropna() # 결측치 삭제한 데이터들을 새로운 데이터프레임에 저장
df2
```

```python
df2.isnull().sum()
```

```python
df.isnull().sum()
```

## 결측데이터 대체하기 --> fillna(채울 값)

```python
df['max_wind'].isnull().sum()
```

```python
df3 = df.copy()  # 복사
df3
```

```python
df3 = df3.fillna(0) # 결측치를 모두 0으로 채웠다
```

```python
df3.isnull().sum()
```

```python
# 평균 --> mean()
avg = df3['max_wind'].mean()
avg
```

```python
df3['max_wind'] = df3['max_wind'].fillna(avg)
```

```python
avg2 = df3['mean_wind'].mean()
avg2
```

```python
df3['mean_wind'] = df3['mean_wind'].fillna(avg2)
```
